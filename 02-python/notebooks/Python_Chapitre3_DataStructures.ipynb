## 1. Vue d'ensemble des structures de données

Une **structure de données** regroupe plusieurs valeurs dans une seule variable.

| Structure | Syntaxe | Ordonnée | Modifiable | Doublons |
|-----------|---------|----------|------------|----------|
| `list`  | `[...]`      | ✅ | ✅ | ✅ |
| `dict`  | `{clé: val}` | ✅ | ✅ | ❌ (clés) |
| `set`   | `{...}`      | ❌ | ✅ | ❌ |
| `tuple` | `(...)`      | ✅ | ❌ | ✅ |

**Analogie :**
- `list` → classeur à pages numérotées
- `dict` → annuaire (nom → numéro)
- `set` → sac de billes uniques
- `tuple` → archive scellée (immuable)


````python
# Aperçu des 4 structures
ma_liste = [1, 2, 3, 4, 5]
mon_dict = {"nom": "Alice", "age": 23}
mon_set  = {1, 2, 3, 3, 3}          # doublons supprimés
mon_tuple = (10, 20, 30)

print("list  :", ma_liste,  "| type :", type(ma_liste).__name__)
print("dict  :", mon_dict,  "| type :", type(mon_dict).__name__)
print("set   :", mon_set,   "| type :", type(mon_set).__name__)
print("tuple :", mon_tuple, "| type :", type(mon_tuple).__name__)
````

## 2. List — Les listes

Une **liste** est une collection **ordonnée** et **modifiable**. La structure la plus utilisée en Python.


````python
# Créer des listes
liste_vide = []
notes      = [14, 16, 12, 18, 15]
etudiants  = ["Alice", "Bob", "Claire", "David"]
profil     = ["Alice", 23, 15.75, True]   # types mixtes

print("Notes     :", notes)
print("Étudiants :", etudiants)
print("Longueur  :", len(notes))
````

### Indexation et slicing

```
fruits = ["mangue", "papaye", "ananas", "banane", "orange"]
            0          1         2         3         4
           -5         -4        -3        -2        -1
```

````python
fruits = ["mangue", "papaye", "ananas", "banane", "orange"]

# Indexation
print("Premier  :", fruits[0])    # mangue
print("Dernier  :", fruits[-1])   # orange
print("3ème     :", fruits[2])    # ananas

# Slicing [debut:fin:pas]  (fin exclue)
print("fruits[1:4] :", fruits[1:4])    # ['papaye', 'ananas', 'banane']
print("fruits[:3]  :", fruits[:3])     # 3 premiers
print("fruits[::-1]:", fruits[::-1])   # liste inversée
````

````python
# Parcourir une liste
etudiants = ["Alice", "Bob", "Claire"]

for i, etudiant in enumerate(etudiants):
    print(f"{i+1}. {etudiant}")
````

## 3. List Manipulation — Manipuler les listes

````python
# AJOUTER des éléments
courses = ["riz", "huile", "sucre"]

courses.append("sel")            # ajoute à la fin
courses.insert(1, "tomates")     # insère à la position 1
courses.extend(["poivre", "cube"]) # fusionne une autre liste

print(courses)
````

````python
# SUPPRIMER des éléments
notes = [14, 16, 12, 14, 18]

notes.remove(14)     # supprime la 1ère occurrence de 14
print("Après remove :", notes)

dernier = notes.pop()  # supprime et retourne le dernier
print("Élément retiré :", dernier, "| Liste :", notes)
````

````python
# TRIER une liste
notes = [14, 18, 12, 16, 10, 15]

notes.sort()
print("Croissant  :", notes)

notes.sort(reverse=True)
print("Décroissant:", notes)
````

````python
# Méthodes statistiques utiles
notes = [14, 16, 12, 18, 14, 15]

print("Somme    :", sum(notes))
print("Min      :", min(notes))
print("Max      :", max(notes))
print("Moyenne  :", sum(notes)/len(notes))
print("Compte 14:", notes.count(14))
print("14 présent ?", 14 in notes)
````

### List Comprehension — la syntaxe élégante

`[expression for element in iterable if condition]`

````python
# Carrés de 1 à 10
carres = [x**2 for x in range(1, 11)]
print("Carrés :", carres)

# Filtrer les notes > 14
notes = [14, 16, 12, 18, 10, 15]
bonnes = [n for n in notes if n > 14]
print("Notes > 14 :", bonnes)
````

## 4. Dictionary — Les dictionnaires

Collection de paires **clé : valeur**. Accès direct via la clé (comme un annuaire).

````python
# Créer un dictionnaire
etudiant = {
    "nom"   : "Alice Dupont",
    "age"   : 23,
    "note"  : 16.5,
    "actif" : True,
    "cours" : ["SQL", "Python", "ML"]
}

print(etudiant)
print("Nombre de clés :", len(etudiant))
````

````python
# Accéder aux valeurs
etudiant = {"nom": "Alice", "age": 23, "ville": "Abidjan"}

print("Par [] :", etudiant["nom"])           # Alice
print("Par get:", etudiant.get("ville"))     # Abidjan
print("Clé absente (get) :", etudiant.get("email", "N/A"))  # N/A (pas d'erreur)
````

````python
# Parcourir un dictionnaire
etudiant = {"nom": "Alice", "age": 23, "note": 16.5}

print("--- Clés + valeurs ---")
for cle, valeur in etudiant.items():
    print(f"{cle} → {valeur}")
````

## 5. Dictionary Manipulation

````python
# AJOUTER / MODIFIER
etudiant = {"nom": "Alice", "age": 23}

etudiant["email"] = "alice@email.com"   # nouvelle clé
etudiant["age"]   = 24                   # modification
etudiant.update({"ville": "Abidjan", "note": 17.5})  # plusieurs à la fois

print(etudiant)
````

````python
# SUPPRIMER
etudiant = {"nom": "Alice", "age": 23, "email": "a@mail.com", "ville": "Abidjan"}

del etudiant["ville"]              # supprimer une clé
age = etudiant.pop("age")          # supprimer et récupérer
print("Âge retiré :", age)
print(etudiant)
````

````python
# Dict comprehension
catalogue = {"Python": 45000, "SQL": 30000, "ML": 60000, "Stats": 35000}

# Filtrer les cours >= 40000
chers = {k: v for k, v in catalogue.items() if v >= 40000}
print("Cours chers :", chers)
````

## 6. Sets — Les ensembles

Collection **non ordonnée** d'éléments **uniques**. Idéal pour supprimer les doublons.

⚠️ Un set vide se crée avec `set()`, pas `{}` (qui crée un dict !).

````python
# Créer un set — les doublons disparaissent
langages = {"Python", "SQL", "R", "Python", "SQL"}
print("Set :", langages)

# Supprimer les doublons d'une liste
notes = [14, 16, 14, 18, 16, 12, 14]
print("Uniques :", set(notes))
````

````python
# Opérations ensemblistes
A = {1, 2, 3, 4, 5}
B = {4, 5, 6, 7, 8}

print("Union        A | B :", A | B)   # tous
print("Intersection A & B :", A & B)   # communs
print("Différence   A - B :", A - B)   # dans A seulement
print("Diff. symét. A ^ B :", A ^ B)   # exclusifs
````

````python
# Cas concret : comparer deux cours
inscrits_sql    = {"Alice", "Bob", "Claire", "David"}
inscrits_python = {"Bob", "David", "Emma", "Fatou"}

print("Suivent les DEUX cours :", inscrits_sql & inscrits_python)
print("SQL mais pas Python    :", inscrits_sql - inscrits_python)
print("Au moins un cours      :", inscrits_sql | inscrits_python)
````

## 7. Tuple — Les tuples

Collection **ordonnée** et **IMMUABLE** (non modifiable). Pour des données constantes.

````python
# Créer des tuples
coordonnees = (5.345, -4.008)     # Abidjan (lat, lon)
rgb_rouge   = (255, 0, 0)
jours       = ("Lun", "Mar", "Mer", "Jeu", "Ven")

print("Coordonnées :", coordonnees)
print("Accès [0]   :", jours[0])
print("Longueur    :", len(jours))

# ⚠️ Tuple d'un seul élément : virgule obligatoire
singleton = (42,)
print("Type singleton :", type(singleton).__name__)
````

````python
# Immuabilité : un tuple ne peut PAS être modifié
point = (3, 5)

try:
    point[0] = 10   # provoque une erreur
except TypeError as e:
    print("Erreur attendue :", e)
````

````python
# Déballage (unpacking)
coordonnees = (5.345, -4.008)
latitude, longitude = coordonnees
print(f"Latitude : {latitude} | Longitude : {longitude}")

# Une fonction qui retourne plusieurs valeurs (= un tuple)
def min_max(liste):
    return min(liste), max(liste)

mini, maxi = min_max([14, 18, 12, 16, 10])
print(f"Min : {mini} | Max : {maxi}")
````

## 8. Structures imbriquées

Le cas le plus courant en Data Science : une **liste de dictionnaires** = un tableau de données.

````python
# Liste de dictionnaires = tableau de données
etudiants = [
    {"nom": "Alice",  "age": 23, "note": 16.5, "ville": "Abidjan"},
    {"nom": "Bob",    "age": 25, "note": 14.0, "ville": "Dakar"},
    {"nom": "Claire", "age": 22, "note": 18.0, "ville": "Accra"},
    {"nom": "David",  "age": 28, "note": 12.5, "ville": "Lagos"},
]

# Accès
print("Nom du 1er   :", etudiants[0]["nom"])
print("Note du 3ème :", etudiants[2]["note"])

print("\n--- Tous les étudiants ---")
for e in etudiants:
    print(f"{e['nom']} ({e['ville']}) — {e['note']}/20")
````

````python
# Analyses sur le tableau de données
notes   = [e["note"] for e in etudiants]
moyenne = sum(notes) / len(notes)
print(f"Moyenne de la promo : {moyenne:.2f}")

# Filtrer les notes >= 15
bons = [e["nom"] for e in etudiants if e["note"] >= 15]
print("Notes >= 15 :", bons)

# Meilleur étudiant
meilleur = max(etudiants, key=lambda e: e["note"])
print(f"Meilleur : {meilleur['nom']} ({meilleur['note']})")
````

````python
# Matrice = liste de listes (rappel du checkpoint Maths !)
D = [
    [4,  8,  12],
    [6,  10, 15],
    [5,  7,  9],
    [3,  6,  9]
]

# Accès [ligne][colonne]
print("D[1][2] =", D[1][2])   # 15

# Somme de chaque colonne
for j in range(len(D[0])):
    total = sum(D[i][j] for i in range(len(D)))
    print(f"Colonne {j} (Produit {chr(65+j)}) — Total : {total}")
````

## 9. Choisir la bonne structure

```
Ordre + modifiable ?        → list  [ ]
Étiquettes clé:valeur ?     → dict  { }
Éviter les doublons ?       → set   { }
Ne doit pas changer ?       → tuple ( )
```

````python
# Conversions entre structures
liste = [1, 2, 2, 3, 3, 4]

print("list → set   :", set(liste))          # supprime doublons
print("list → tuple :", tuple(liste))        # rend immuable

d = {"a": 1, "b": 2, "c": 3}
print("dict → clés  :", list(d.keys()))
print("dict → valeurs:", list(d.values()))
````
