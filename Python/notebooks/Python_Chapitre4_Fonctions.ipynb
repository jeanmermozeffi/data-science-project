## 1. Qu'est-ce qu'une fonction ?

Une **fonction** est un bloc de code réutilisable qui exécute une tâche précise.

**Analogie :** une fonction, c'est comme une machine à café — vous donnez des entrées, elle exécute un processus, elle rend une sortie.

Vous utilisez déjà des fonctions natives : `print()`, `len()`, `sum()`, `type()`...


````python
# Fonctions déjà connues
print("Bonjour")
print(len([1, 2, 3]))
print(sum([1, 2, 3]))
print(type(42))
````

## 4. Fonctions à plusieurs paramètres

````python
def moyenne_ponderee(note_examen, note_projet, note_partiel):
    """50% examen, 30% projet, 20% partiel"""
    return (note_examen * 0.5) + (note_projet * 0.3) + (note_partiel * 0.2)

print(moyenne_ponderee(16, 18, 14))   # 16.2
````

````python
def evaluer_note(nom, note):
    if note >= 16:
        mention = "Très bien 🏆"
    elif note >= 14:
        mention = "Bien 👍"
    elif note >= 10:
        mention = "Passable"
    else:
        mention = "Insuffisant ❌"
    return f"{nom} : {note}/20 — {mention}"

print(evaluer_note("Alice", 17))
print(evaluer_note("Bob", 8))
````

## 6. L'instruction de retour (return)

Sans `return`, une fonction renvoie `None` par défaut.

````python
# Sans return
def afficher_carre(n):
    print(n ** 2)

resultat = afficher_carre(5)   # Affiche 25
print("Valeur retournée :", resultat)   # None !
````

````python
# return arrête immédiatement la fonction
def verifier_age(age):
    if age < 0:
        return "Âge invalide"
    if age < 18:
        return "Mineur"
    return "Majeur"

print(verifier_age(-5))
print(verifier_age(15))
print(verifier_age(25))
````

````python
# Retourner plusieurs valeurs (tuple)
def statistiques(notes):
    return min(notes), max(notes), sum(notes)/len(notes)

notes = [14, 18, 12, 16, 10]
mini, maxi, moy = statistiques(notes)
print(f"Min : {mini}, Max : {maxi}, Moyenne : {moy:.2f}")
````

## 7. \*args et \*\*kwargs — Paramètres flexibles

- `*args` → arguments positionnels illimités (regroupés en **tuple**)
- `**kwargs` → arguments nommés illimités (regroupés en **dict**)

````python
def somme(*args):
    print(f"args reçus : {args} (type: {type(args).__name__})")
    return sum(args)

print(somme(1, 2))
print(somme(1, 2, 3, 4, 5))
````

````python
def creer_profil_flex(**kwargs):
    print(f"kwargs reçus : {kwargs}")
    for cle, valeur in kwargs.items():
        print(f"  {cle} : {valeur}")

creer_profil_flex(nom="Alice", age=23, ville="Abidjan")
````

````python
# Cas d'usage : statistiques flexibles
def statistiques_flexibles(*nombres):
    if not nombres:
        return "Aucune donnée fournie"
    return {
        "somme"   : sum(nombres),
        "moyenne" : sum(nombres) / len(nombres),
        "min"     : min(nombres),
        "max"     : max(nombres)
    }

print(statistiques_flexibles(14, 16, 18))
print(statistiques_flexibles(10, 20, 30, 40, 50))
````

## 8. Portée des variables (Scope)

Une variable créée **dans** une fonction (locale) n'existe pas en dehors.

````python
x = 10   # variable GLOBALE

def ma_fonction():
    y = 5   # variable LOCALE
    print(f"À l'intérieur : x = {x}, y = {y}")

ma_fonction()
print("x en dehors :", x)

try:
    print(y)   # provoque une erreur
except NameError as e:
    print("Erreur attendue :", e)
````

````python
# Modifier une variable globale avec 'global'
compteur = 0

def incrementer():
    global compteur
    compteur += 1

incrementer()
incrementer()
incrementer()
print("Compteur final :", compteur)
````

## 9. Fonction Lambda

Fonction courte, anonyme, définie en une seule ligne :
```python
lambda paramètres: expression
```

````python
# def vs lambda
def carre(x):
    return x ** 2

carre_lambda = lambda x: x ** 2

print(carre(5))
print(carre_lambda(5))
````

````python
addition   = lambda a, b: a + b
moyenne    = lambda a, b, c: (a + b + c) / 3
est_pair   = lambda n: n % 2 == 0

print(addition(3, 5))
print(moyenne(14, 16, 18))
print(est_pair(4), est_pair(7))
````

````python
# Cas d'usage typique : trier avec une clé lambda
etudiants = [
    {"nom": "Alice",  "note": 16},
    {"nom": "Bob",    "note": 12},
    {"nom": "Claire", "note": 18}
]

etudiants_tries = sorted(etudiants, key=lambda e: e["note"], reverse=True)
for e in etudiants_tries:
    print(f"{e['nom']} : {e['note']}")
````

## 10. Fonction Map

`map(fonction, iterable)` applique une fonction à **chaque élément** d'une séquence.

````python
notes = [14, 16, 12, 18, 10]

# Avec lambda
notes_sur_10 = list(map(lambda n: n / 2, notes))
print("Notes /2 :", notes_sur_10)

# Avec fonction classique
def convertir_en_pourcentage(note):
    return (note / 20) * 100

pourcentages = list(map(convertir_en_pourcentage, notes))
print("Pourcentages :", pourcentages)
````

````python
# map() vs list comprehension — équivalents
notes = [14, 16, 12, 18]

resultat_map = list(map(lambda n: n * 2, notes))
resultat_comprehension = [n * 2 for n in notes]

print("map()             :", resultat_map)
print("list comprehension:", resultat_comprehension)
````

````python
# map() sur plusieurs listes
prix      = [1000, 2500, 500]
quantites = [3, 2, 10]

totaux = list(map(lambda p, q: p * q, prix, quantites))
print("Totaux :", totaux)
````

## 11. Filter et Reduce

- `filter()` garde les éléments où la fonction retourne `True`
- `reduce()` réduit une séquence à une seule valeur (module `functools`)

````python
notes = [8, 14, 16, 9, 18, 11, 20]

# filter() : garder les notes >= 10
notes_admises = list(filter(lambda n: n >= 10, notes))
print("Notes admises :", notes_admises)
````

````python
from functools import reduce

nombres = [1, 2, 3, 4, 5]

# reduce() : produit de tous les nombres
produit = reduce(lambda a, b: a * b, nombres)
print("Produit :", produit)   # 120

# reduce() : trouver le maximum
maximum = reduce(lambda a, b: a if a > b else b, nombres)
print("Maximum :", maximum)
````

````python
# Pipeline complet : filter → map → reduce
from functools import reduce

notes = [8, 14, 16, 9, 18, 11, 20]

notes_admises = list(filter(lambda n: n >= 10, notes))
notes_sur_100 = list(map(lambda n: n * 5, notes_admises))
total         = reduce(lambda a, b: a + b, notes_sur_100)

print(f"Notes admises : {notes_admises}")
print(f"Sur 100       : {notes_sur_100}")
print(f"Total         : {total}")
````

## 12. Fonctions et Data Science

Applications concrètes des fonctions dans un contexte data.

````python
# Nettoyage de données
def nettoyer_texte(texte):
    return texte.strip().lower()

noms_bruts = ["  Alice  ", "BOB", " Claire"]
noms_propres = list(map(nettoyer_texte, noms_bruts))
print(noms_propres)
````

````python
# Normalisation min-max (utile pour le Machine Learning)
def normaliser(valeur, minimum, maximum):
    return (valeur - minimum) / (maximum - minimum)

notes = [8, 12, 15, 18, 20]
mini, maxi = min(notes), max(notes)

notes_normalisees = [round(normaliser(n, mini, maxi), 2) for n in notes]
print(notes_normalisees)
````

````python
# Fonction de résumé statistique réutilisable
def resume_statistique(donnees):
    n = len(donnees)
    moyenne = sum(donnees) / n
    variance = sum((x - moyenne) ** 2 for x in donnees) / n
    ecart_type = variance ** 0.5
    return {
        "n": n, "moyenne": round(moyenne, 2),
        "min": min(donnees), "max": max(donnees),
        "variance": round(variance, 2), "ecart_type": round(ecart_type, 2)
    }

ventes = [4500, 7800, 11200, 6300, 9100]
print(resume_statistique(ventes))
````

````python
# Pipeline de traitement de données (charger → nettoyer → filtrer)
def charger_donnees():
    return [
        {"nom": " alice ", "note": 16.234},
        {"nom": "BOB",     "note": 12.876},
        {"nom": " Claire", "note": 18.512}
    ]

def nettoyer_donnees(donnees):
    for d in donnees:
        d["nom"]  = d["nom"].strip().title()
        d["note"] = round(d["note"], 1)
    return donnees

def filtrer_admis(donnees):
    return [d for d in donnees if d["note"] >= 14]

donnees = charger_donnees()
donnees = nettoyer_donnees(donnees)
admis   = filtrer_admis(donnees)

for etudiant in admis:
    print(etudiant)
````

````python
# Bonus : fonction récursive
def factorielle(n):
    if n <= 1:
        return 1
    return n * factorielle(n - 1)

print("5! =", factorielle(5))

def fibonacci(n):
    if n <= 1:
        return n
    return fibonacci(n - 1) + fibonacci(n - 2)

print("Fibonacci :", [fibonacci(i) for i in range(8)])
````

### 🏆 Challenge bonus — Pipeline de traitement

Complétez le pipeline de fonctions pour analyser les ventes d'un bootcamp :
1. `charger_ventes()` — retourne les données brutes
2. `calculer_totaux(ventes)` — ajoute une clé `"total"`
3. `filtrer_grosses_ventes(ventes, seuil)` — filtre selon un seuil
4. `chiffre_affaires_total(ventes)` — utilise `reduce` pour le CA total

````python
from functools import reduce

ventes_brutes = [
    {"produit": "Formation SQL",    "quantite": 15, "prix_unitaire": 25000},
    {"produit": "Formation Python", "quantite": 20, "prix_unitaire": 35000},
    {"produit": "Formation ML",     "quantite": 8,  "prix_unitaire": 50000},
]

# ✏️ À vous de jouer !
````

````python
# ✅ Solution Challenge
from functools import reduce

def charger_ventes():
    return [
        {"produit": "Formation SQL",    "quantite": 15, "prix_unitaire": 25000},
        {"produit": "Formation Python", "quantite": 20, "prix_unitaire": 35000},
        {"produit": "Formation ML",     "quantite": 8,  "prix_unitaire": 50000},
    ]

def calculer_totaux(ventes):
    for v in ventes:
        v["total"] = v["quantite"] * v["prix_unitaire"]
    return ventes

def filtrer_grosses_ventes(ventes, seuil):
    return [v for v in ventes if v["total"] >= seuil]

def chiffre_affaires_total(ventes):
    return reduce(lambda acc, v: acc + v["total"], ventes, 0)

ventes = charger_ventes()
ventes = calculer_totaux(ventes)
grosses_ventes = filtrer_grosses_ventes(ventes, 400000)
ca_total = chiffre_affaires_total(ventes)

print("--- Toutes les ventes ---")
for v in ventes:
    print(v)

print("\n--- Grosses ventes (≥ 400 000) ---")
for v in grosses_ventes:
    print(v)

print(f"\nChiffre d'affaires total : {ca_total} FCFA")
````
