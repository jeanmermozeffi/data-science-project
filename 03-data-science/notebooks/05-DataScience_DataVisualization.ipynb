---
## 1. Introduction — Charger et nettoyer les données

On visualise toujours des données **propres** (compétences des chapitres précédents).

````python
df = pd.read_csv("bootcamp_500.csv")

# Nettoyage rapide
df["ville"] = df["ville"].str.strip().str.title()
df.loc[(df["age"] < 15) | (df["age"] > 100), "age"] = np.nan
df.loc[df["note_sql"] > 20, "note_sql"] = np.nan
for col in ["age", "note_sql", "note_python"]:
    df[col] = df[col].fillna(df[col].median())
df["ville"] = df["ville"].fillna(df["ville"].mode()[0])
df = df.drop_duplicates().reset_index(drop=True)

print("Données nettoyées :", df.shape)
df.head()
````

---
## 2. Exploring Popular Chart Types

### 2.1 Histogramme — Distribution d'une variable

````python
plt.figure(figsize=(8, 5))
plt.hist(df["note_sql"], bins=20, color="#3498db", edgecolor="white")
plt.title("Distribution des notes SQL")
plt.xlabel("Note SQL")
plt.ylabel("Nombre d'étudiants")
plt.show()
````

### 2.2 Diagramme en barres — Comparer des catégories

````python
comptage = df["ville"].value_counts()

plt.figure(figsize=(10, 5))
comptage.plot(kind="bar", color="#2ecc71")
plt.title("Nombre d'étudiants par ville")
plt.xlabel("Ville"); plt.ylabel("Nombre d'étudiants")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
````

### 2.3 Nuage de points — Relation entre deux variables

````python
plt.figure(figsize=(8, 5))
plt.scatter(df["heures_etude"], df["note_sql"], alpha=0.5, color="#e74c3c")
plt.title("Relation heures d'étude / note SQL")
plt.xlabel("Heures d'étude par semaine")
plt.ylabel("Note SQL")
plt.show()
# → tendance montante nette : plus on étudie, meilleure est la note
````

### 2.4 Boxplot — Distribution par groupe

````python
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x="niveau", y="note_sql", order=["Débutant","Intermédiaire","Avancé"])
plt.title("Notes SQL par niveau")
plt.show()
````

### 2.5 Camembert — Proportions (à utiliser avec parcimonie)

````python
proportions = df["niveau"].value_counts()

plt.figure(figsize=(7, 7))
plt.pie(proportions, labels=proportions.index, autopct="%1.1f%%",
        colors=["#3498db", "#2ecc71", "#e74c3c"])
plt.title("Répartition des niveaux")
plt.show()
````

### 2.6 Heatmap — Matrice de corrélation

````python
correlation = df[["age","heures_etude","note_sql","note_python","salaire_stage"]].corr()

plt.figure(figsize=(8, 6))
sns.heatmap(correlation, annot=True, cmap="coolwarm", center=0, fmt=".2f")
plt.title("Matrice de corrélation")
plt.show()
````

---
## 3. How to Choose the Right Chart

```
Distribution d'1 variable numérique → HISTOGRAMME / BOXPLOT
Fréquence d'1 catégorie             → BARRES
Relation 2 variables numériques     → NUAGE DE POINTS
Évolution temporelle                → COURBE
Distribution par groupe             → BOXPLOT / VIOLIN
Corrélations multiples              → HEATMAP
```

---
## 4. Matplotlib & Seaborn

### 4.1 Subplots — Plusieurs graphiques côte à côte (Matplotlib)

````python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].hist(df["note_sql"], bins=20, color="#3498db")
axes[0].set_title("Distribution note SQL")
axes[0].set_xlabel("Note")

axes[1].scatter(df["heures_etude"], df["note_sql"], alpha=0.5, color="#e74c3c")
axes[1].set_title("Heures d'étude vs note")
axes[1].set_xlabel("Heures")

plt.tight_layout()
plt.show()
````

### 4.2 Seaborn — Histogramme avec densité

````python
plt.figure(figsize=(8, 5))
sns.histplot(data=df, x="note_sql", kde=True, color="#3498db")
plt.title("Distribution des notes SQL (avec densité)")
plt.show()
````

### 4.3 Seaborn — Nuage avec régression et couleur par groupe

````python
sns.lmplot(data=df, x="heures_etude", y="note_sql", height=5, aspect=1.5)
plt.title("Tendance : heures d'étude → note SQL")
plt.show()
````

### 4.4 Seaborn — Violin plot

````python
plt.figure(figsize=(8, 5))
sns.violinplot(data=df, x="niveau", y="note_python", order=["Débutant","Intermédiaire","Avancé"])
plt.title("Distribution des notes Python par niveau")
plt.show()
````

### 4.5 Seaborn — Pairplot (toutes les relations d'un coup)

````python
sns.pairplot(df[["heures_etude","note_sql","note_python","niveau"]], hue="niveau")
plt.show()
````

## 5. Plotly — Graphiques interactifs

Survolez les points, zoomez, masquez des séries !

````python
fig = px.scatter(
    df, x="heures_etude", y="note_sql", color="niveau",
    hover_data=["prenom", "ville"],
    title="Heures d'étude vs Note SQL (interactif)"
)
fig.show()
````

````python
fig = px.histogram(df, x="note_sql", nbins=20, color="niveau",
                   title="Distribution des notes SQL par niveau")
fig.show()
````

````python
fig = px.box(df, x="niveau", y="note_python",
             title="Notes Python par niveau (interactif)")
fig.show()
````

````python
# Graphique 3D interactif
fig = px.scatter_3d(
    df, x="heures_etude", y="note_sql", z="note_python",
    color="niveau", title="Vue 3D : heures, note SQL, note Python"
)
fig.show()
````

## 6. Choisir entre Plotly, Matplotlib et Seaborn

| Besoin | Choix |
|--------|-------|
| Exploration rapide, graphiques statistiques | **Seaborn** |
| Contrôle total, publication | **Matplotlib** |
| Interactivité, dashboards web | **Plotly** |

Le même graphique dans les trois :

````python
# Matplotlib
plt.figure(figsize=(6,4))
plt.scatter(df["heures_etude"], df["note_sql"], alpha=0.5)
plt.title("Matplotlib"); plt.show()

# Seaborn
plt.figure(figsize=(6,4))
sns.scatterplot(data=df, x="heures_etude", y="note_sql", hue="niveau")
plt.title("Seaborn"); plt.show()

# Plotly
fig = px.scatter(df, x="heures_etude", y="note_sql", color="niveau", title="Plotly")
fig.show()
````

## 7. Bonnes pratiques

Un bon graphique : titre clair, axes étiquetés, axes honnêtes, couleurs utiles, accessible.

````python
fig, ax = plt.subplots(figsize=(10, 6))
sns.barplot(data=df, x="ville", y="note_sql", ax=ax,
            estimator="mean", errorbar=None, palette="viridis")
ax.set_title("Note SQL moyenne par ville", fontsize=14, fontweight="bold")
ax.set_xlabel("Ville", fontsize=12)
ax.set_ylabel("Note SQL moyenne", fontsize=12)
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig("note_par_ville.png", dpi=150, bbox_inches="tight")
plt.show()
print("✅ Graphique sauvegardé en haute qualité")
````

## 8. Python Project — Dashboard du Bootcamp

Un tableau de bord 2x2 qui raconte l'histoire des données.

````python
df["moyenne"] = (df["note_sql"] + df["note_python"]) / 2

fig, axes = plt.subplots(2, 2, figsize=(16, 12))
fig.suptitle("📊 Dashboard du Bootcamp Data Science", fontsize=18, fontweight="bold")

# 1. Distribution des moyennes
sns.histplot(data=df, x="moyenne", bins=20, kde=True, ax=axes[0,0], color="#3498db")
axes[0,0].set_title("Distribution des moyennes")

# 2. Étudiants par ville
comptage = df["ville"].value_counts()
sns.barplot(x=comptage.values, y=comptage.index, ax=axes[0,1], palette="viridis")
axes[0,1].set_title("Nombre d'étudiants par ville")

# 3. Heures d'étude vs moyenne
sns.scatterplot(data=df, x="heures_etude", y="moyenne", hue="niveau", alpha=0.6, ax=axes[1,0])
axes[1,0].set_title("Relation heures d'étude / moyenne")

# 4. Moyenne par niveau
sns.boxplot(data=df, x="niveau", y="moyenne", order=["Débutant","Intermédiaire","Avancé"], ax=axes[1,1])
axes[1,1].set_title("Moyenne par niveau")

plt.tight_layout()
plt.savefig("dashboard_bootcamp.png", dpi=150, bbox_inches="tight")
plt.show()
print("✅ Dashboard sauvegardé dans dashboard_bootcamp.png")
````

````python
# Bonus : dashboard interactif Plotly
fig = px.scatter(
    df, x="heures_etude", y="moyenne", color="niveau",
    size="salaire_stage", hover_data=["prenom", "ville"],
    title="Dashboard interactif — Heures d'étude vs Moyenne"
)
fig.write_html("dashboard_interactif.html")
fig.show()
print("✅ Dashboard interactif sauvegardé dans dashboard_interactif.html")
````

````python
# ✅ Piste de solution
df["moyenne"] = (df["note_sql"] + df["note_python"]) / 2

fig, axes = plt.subplots(2, 2, figsize=(15, 11))

sns.histplot(df["moyenne"], bins=20, kde=True, ax=axes[0,0], color="#3498db")
axes[0,0].set_title("1. Qui réussit ?")

sns.regplot(data=df, x="heures_etude", y="moyenne", ax=axes[0,1], scatter_kws={"alpha":0.4})
axes[0,1].set_title("2. L'effort paie-t-il ?")

sns.boxplot(data=df, x="niveau", y="moyenne", order=["Débutant","Intermédiaire","Avancé"], ax=axes[1,0])
axes[1,0].set_title("3. Réussite par niveau")

corr = df[["heures_etude","note_sql","note_python","moyenne"]].corr()
sns.heatmap(corr, annot=True, cmap="coolwarm", center=0, fmt=".2f", ax=axes[1,1])
axes[1,1].set_title("4. Corrélations")

plt.tight_layout()
plt.show()

print("CONCLUSION : forte corrélation (~0.8) entre heures d'étude et moyenne")
print("→ l'effort est le principal facteur de réussite observé.")
````
