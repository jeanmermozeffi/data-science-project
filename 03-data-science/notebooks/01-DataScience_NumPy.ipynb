## 2. Pourquoi NumPy et pas des listes Python ?

NumPy est généralement **10 à 100 fois plus rapide** grâce à la **vectorisation** : les opérations s'exécutent en code C optimisé, sans boucle Python explicite.

````python
import time

# Liste Python classique
liste = list(range(1_000_000))
debut = time.time()
liste_carre = [x**2 for x in liste]
temps_liste = time.time() - debut

# Tableau NumPy
tableau = np.arange(1_000_000)
debut = time.time()
tableau_carre = tableau ** 2
temps_numpy = time.time() - debut

print(f"Liste Python  : {temps_liste:.4f} secondes")
print(f"Tableau NumPy : {temps_numpy:.4f} secondes")
print(f"NumPy est ~{temps_liste/temps_numpy:.0f}x plus rapide !")
````

````python
# Vectorisation : pas besoin de boucle avec NumPy
notes = [14, 16, 12, 18, 10]

# Liste Python — boucle nécessaire
notes_sur_100 = [n * 5 for n in notes]
print("Liste :", notes_sur_100)

# NumPy — opération vectorisée
notes_np = np.array([14, 16, 12, 18, 10])
notes_sur_100_np = notes_np * 5
print("NumPy :", notes_sur_100_np)
````

## 5. Types de données NumPy (dtype)

⚠️ Contrairement à une liste Python, **tous les éléments d'un tableau NumPy ont le même type**.

````python
entiers  = np.array([1, 2, 3])
decimaux = np.array([1.5, 2.5, 3.5])
mixte    = np.array([1, 2, 3.5])   # converti automatiquement en float

print("entiers.dtype :", entiers.dtype)
print("decimaux.dtype:", decimaux.dtype)
print("mixte         :", mixte, "| dtype:", mixte.dtype)

# Conversion avec astype
convertis = np.array([1.9, 2.7, 3.1]).astype(int)
print("astype(int) :", convertis, "  (troncature, pas arrondi !)")
````

## 6. Array Reshaping — Remodeler un tableau

`reshape()` change la forme sans changer les données (le nombre total d'éléments doit rester identique).

````python
donnees = np.arange(12)
print("Original :", donnees.shape)

matrice = donnees.reshape(3, 4)
print(matrice)

# -1 : NumPy calcule automatiquement cette dimension
matrice2 = donnees.reshape(3, -1)
print("Shape avec -1 :", matrice2.shape)
````

````python
# Aplatir et transposer
matrice = np.array([[1, 2, 3], [4, 5, 6]])

print("flatten :", matrice.flatten())
print("Transposée :\n", matrice.T)
print("Shape original :", matrice.shape, "| Shape transposée :", matrice.T.shape)
````

## 7. Array Indexing — Accéder aux éléments

````python
notes = np.array([14, 16, 12, 18, 15])
print(notes[0], notes[2], notes[-1])
````

````python
ventes = np.array([
    [4,  8,  12],   # Région 1
    [6,  10, 15],   # Région 2
    [5,  7,  9]     # Région 3
])

print("ventes[0, 0] :", ventes[0, 0])    # Région 1, Produit A
print("ventes[1, 2] :", ventes[1, 2])    # Région 2, Produit C
print("Ligne 0      :", ventes[0])        # toute la Région 1
print("Colonne 0    :", ventes[:, 0])     # tout le Produit A
````

## 8. Array Slicing — Découper un tableau

⚠️ **Le slicing crée une VUE, pas une copie !**

````python
notes = np.array([14, 16, 12, 18, 15, 10, 19])

print(notes[1:4])
print(notes[:3])
print(notes[::2])
print(notes[::-1])
````

````python
# ⚠️ Piège : le slicing partage la mémoire avec l'original
original = np.array([1, 2, 3, 4, 5])
extrait  = original[1:4]
extrait[0] = 999

print("extrait  :", extrait)
print("original :", original, "← a changé aussi !")

# ✅ Pour éviter ça : .copy()
original2 = np.array([1, 2, 3, 4, 5])
copie = original2[1:4].copy()
copie[0] = 999
print("original2 (inchangé) :", original2)
````

## 9. Indexation booléenne et Fancy Indexing

L'une des fonctionnalités les plus puissantes de NumPy : filtrer via une **condition**.

````python
notes = np.array([8, 14, 16, 9, 18, 11, 20])

masque = notes >= 10
print("Masque :", masque)
print("Notes admises :", notes[masque])

# Combiner des conditions : & (et), | (ou)
print("Entre 12 et 18 :", notes[(notes >= 12) & (notes <= 18)])
````

````python
# Modifier avec un masque
notes2 = np.array([8, 14, 16, 9, 18])
notes2[notes2 < 10] = 0
print("Après modification :", notes2)

# np.where() : condition vectorisée
resultat = np.where(notes2 >= 10, "Admis", "Ajourné")
print(resultat)
````

````python
# Fancy indexing : sélectionner via une liste d'indices
notes = np.array([14, 16, 12, 18, 10, 20, 8])
indices = [0, 2, 4]
print(notes[indices])
````

## 10. NumPy Random — Génération aléatoire

````python
print("rand(5)        :", np.random.rand(5))
print("randint(1,21,10):", np.random.randint(1, 21, size=10))
print("randn(5)        :", np.random.randn(5))
````

````python
etudiants = np.array(["Alice", "Bob", "Claire", "David", "Emma"])

print("Choix aléatoire  :", np.random.choice(etudiants))
print("3 sans remise    :", np.random.choice(etudiants, size=3, replace=False))
````

````python
# seed() : rendre l'aléatoire REPRODUCTIBLE (essentiel en Data Science)
np.random.seed(42)
print("1er tirage :", np.random.rand(3))

np.random.seed(42)   # même graine
print("2e tirage  :", np.random.rand(3), "← identique au premier !")
````

## 11. NumPy Addition/Soustraction et le Broadcasting

Les opérations s'appliquent **élément par élément**. Le **broadcasting** permet d'opérer entre tableaux de formes différentes.

````python
a = np.array([1, 2, 3])
b = np.array([10, 20, 30])

print("a + b :", a + b)
print("a * b :", a * b)

# ⚠️ Comparaison avec les listes Python
print("Listes Python [1,2,3]+[10,20,30] :", [1,2,3] + [10,20,30], "← CONCATÉNATION !")
````

````python
# Broadcasting : scalaire sur toute la matrice
ventes = np.array([[4, 8, 12], [6, 10, 15], [5, 7, 9]])
print("ventes + 100 :\n", ventes + 100)

# Broadcasting : vecteur ligne sur chaque ligne
prime_par_produit = np.array([1, 2, 3])
print("\nventes + prime :\n", ventes + prime_par_produit)
````

## 12. Fonctions d'agrégation

Le concept d'**axis** est essentiel : `axis=0` agit verticalement (par colonne), `axis=1` horizontalement (par ligne).

````python
notes = np.array([14, 16, 12, 18, 10])

print("sum   :", notes.sum())
print("mean  :", notes.mean())
print("std   :", notes.std())
print("min/max :", notes.min(), notes.max())
print("argmin/argmax (indices) :", notes.argmin(), notes.argmax())
````

````python
ventes = np.array([
    [4,  8,  12],   # Région 1
    [6,  10, 15],   # Région 2
    [5,  7,  9]     # Région 3
])

print("Total par produit (axis=0) :", ventes.sum(axis=0))
print("Total par région  (axis=1) :", ventes.sum(axis=1))
````

## 13. Some NumPy Functions — Fonctions utiles

````python
tableau = np.array([1, 4, 9, 16, 25])

print("sqrt  :", np.sqrt(tableau))
print("round :", np.round([3.456, 7.891], 1))
print("abs   :", np.abs([-5, 3, -2]))
````

## 14. Algèbre linéaire avec NumPy

On retrouve directement les opérations matricielles du checkpoint de mathématiques !

````python
D = np.array([
    [4,  8,  12],
    [6,  10, 15],
    [5,  7,  9],
    [3,  6,  9]
])
P = np.array([5, 10, 15])   # vecteur de prix

# Produit matriciel — comme au checkpoint Maths !
revenus = D @ P
print("Revenus par région :", revenus)

# Transposée
print("Transposée :\n", D.T)
````

````python
A = np.array([[1, 2], [3, 4]])

print("Déterminant :", np.linalg.det(A))
print("Inverse :\n", np.linalg.inv(A))
print("Trace :", np.trace(A))

# Résoudre Ax = b
b = np.array([5, 6])
x = np.linalg.solve(A, b)
print("Solution x :", x)
````

### 🏆 Challenge bonus — Reprendre le checkpoint Maths avec NumPy

Avec `D = [[4,8,12],[6,10,15],[5,7,9],[3,6,9]]` et `P = [5,10,15]` :
1. Créez D et P avec NumPy
2. Calculez moyenne, médiane, écart type de chaque colonne
3. Calculez la transposée de D
4. Calculez les revenus par région (`D @ P`)
5. Trouvez la région au revenu maximal (`np.argmax()`)

````python
# ✏️ À vous de jouer !

````

````python
# ✅ Solution Challenge
D = np.array([[4, 8, 12], [6, 10, 15], [5, 7, 9], [3, 6, 9]])
P = np.array([5, 10, 15])

print("Moyenne :", D.mean(axis=0))
print("Médiane :", np.median(D, axis=0))
print("Écart type :", D.std(axis=0))

print("Transposée :\n", D.T)

revenus = D @ P
print("Revenus par région :", revenus)

region_max = np.argmax(revenus)
print(f"Région avec le + gros revenu : Région {region_max + 1} ({revenus[region_max]})")
````
